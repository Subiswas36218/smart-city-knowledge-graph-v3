```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)
```

```python
from src.cleaning import *

print("✓ src.cleaning imported successfully")
```

```python
import geopandas as gpd

from src.cleaning import (
    clean_pois,
    clean_buildings,
    clean_roads,
    clean_transport,
)

RAW_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

PROCESSED_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

# ------------------------------------------------------------
# Load raw datasets
# ------------------------------------------------------------

pois_raw = gpd.read_parquet(
    RAW_DIR / "pois_raw.parquet"
)

buildings_raw = gpd.read_parquet(
    RAW_DIR / "buildings_raw.parquet"
)

roads_raw = gpd.read_parquet(
    RAW_DIR / "roads_raw.parquet"
)

transport_raw = gpd.read_parquet(
    RAW_DIR / "transport_raw.parquet"
)

print("Raw datasets")
print("-" * 50)
print(f"POIs:        {len(pois_raw):,}")
print(f"Buildings:   {len(buildings_raw):,}")
print(f"Roads:       {len(roads_raw):,}")
print(f"Transport:   {len(transport_raw):,}")
```

```python
# ------------------------------------------------------------
# Clean datasets
# ------------------------------------------------------------

pois_clean = clean_pois(pois_raw)

buildings_clean = clean_buildings(
    buildings_raw
)

roads_clean = clean_roads(
    roads_raw
)

transport_clean = clean_transport(
    transport_raw
)

print("\nCleaned datasets")
print("-" * 50)
print(f"POIs:        {len(pois_clean):,}")
print(f"Buildings:   {len(buildings_clean):,}")
print(f"Roads:       {len(roads_clean):,}")
print(f"Transport:   {len(transport_clean):,}")
```

```python
# ------------------------------------------------------------
# Save processed datasets
# ------------------------------------------------------------

pois_clean.to_parquet(
    PROCESSED_DIR / "pois_clean.parquet",
    index=False,
)

buildings_clean.to_parquet(
    PROCESSED_DIR / "buildings_clean.parquet",
    index=False,
)

roads_clean.to_parquet(
    PROCESSED_DIR / "roads_clean.parquet",
    index=False,
)

transport_clean.to_parquet(
    PROCESSED_DIR / "transport_clean.parquet",
    index=False,
)

print("\n✓ All cleaned datasets saved.")
```

```python
for filename in [
    "pois_clean.parquet",
    "buildings_clean.parquet",
    "roads_clean.parquet",
    "transport_clean.parquet",
]:
    path = PROCESSED_DIR / filename

    print(
        f"{'✓' if path.exists() else '✗'} "
        f"{filename}"
    )
```
