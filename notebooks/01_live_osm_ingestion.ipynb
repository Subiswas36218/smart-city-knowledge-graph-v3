# 01 — Live OSM ingestion


```python
from pathlib import Path
import sys

PROJECT_ROOT = Path(
    "/Users/subhankarbiswas/smart-city-knowledge-graph-v3"
)

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

RAW_DIR = PROJECT_ROOT / "data" / "raw"
RAW_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

CITY = "Eschwege, Hesse, Germany"

print("Project:", PROJECT_ROOT)
print("City:", CITY)
print("Raw directory:", RAW_DIR)
```

```python
import osmnx as ox

from src.osm_ingestion import (
    fetch_pois,
    fetch_buildings,
    fetch_roads,
    fetch_transport,
    fetch_walk_network,
)

print("=" * 60)
print("OSM RAW DATA INGESTION")
print("=" * 60)
```

```python
print("\n[1/6] Downloading POIs...")

pois = fetch_pois(CITY)

print(f"✓ {len(pois):,} POIs downloaded")

from src.osm_ingestion import save_geodata

save_geodata(
    pois,
    RAW_DIR / "pois_raw.parquet",
)

print("✓ pois_raw.parquet saved")
```

```python
print("\n[2/6] Downloading buildings...")

buildings = fetch_buildings(CITY)

print(f"✓ {len(buildings):,} buildings downloaded")

save_geodata(
    buildings,
    RAW_DIR / "buildings_raw.parquet",
)

print("✓ buildings_raw.parquet saved")
```

```python
print("\n[3/6] Downloading roads...")

roads = fetch_roads(CITY)

print(f"✓ {len(roads):,} road features downloaded")

save_geodata(
    roads,
    RAW_DIR / "roads_raw.parquet",
)

print("✓ roads_raw.parquet saved")
```

```python
print("\n[4/6] Downloading transport...")

transport = fetch_transport(CITY)

print(f"✓ {len(transport):,} transport features downloaded")

save_geodata(
    transport,
    RAW_DIR / "transport_raw.parquet",
)

print("✓ transport_raw.parquet saved")
```

```python
print("\n[5/6] Downloading walking network...")

walk = fetch_walk_network(CITY)

ox.save_graphml(
    walk,
    RAW_DIR / "walk_network.graphml",
)

print(
    f"✓ Walking network: "
    f"{walk.number_of_nodes():,} nodes / "
    f"{walk.number_of_edges():,} edges"
)
```

```python
print("\n[6/6] Downloading driving network...")

drive = ox.graph_from_place(
    CITY,
    network_type="drive",
    simplify=True,
)

ox.save_graphml(
    drive,
    RAW_DIR / "drive_network.graphml",
)

print(
    f"✓ Driving network: "
    f"{drive.number_of_nodes():,} nodes / "
    f"{drive.number_of_edges():,} edges"
)
```

```python
print("\n" + "=" * 60)
print("RAW DATA VERIFICATION")
print("=" * 60)

required = [
    "pois_raw.parquet",
    "buildings_raw.parquet",
    "roads_raw.parquet",
    "transport_raw.parquet",
    "walk_network.graphml",
    "drive_network.graphml",
]

for filename in required:

    path = RAW_DIR / filename

    if path.exists():
        size_mb = path.stat().st_size / (1024 * 1024)
        print(
            f"✓ {filename:<30} "
            f"{size_mb:.2f} MB"
        )
    else:
        print(f"✗ MISSING: {filename}")
```
