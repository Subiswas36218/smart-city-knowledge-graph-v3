# 03 — OSM ↔ Wikidata reconciliation


```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)
```

```python
import geopandas as gpd

from src.config import settings
from src.wikidata import enrich_osm_with_wikidata

POIS_PATH = PROJECT_ROOT / "data" / "processed" / "pois_clean.parquet"
OUTPUT_PATH = PROJECT_ROOT / "data" / "processed" / "pois_wikidata.parquet"

pois = gpd.read_parquet(POIS_PATH)

print(f"POIs available for Wikidata enrichment: {len(pois):,}")
print(f"Wikidata enabled: {settings.wikidata_enabled}")
```

```python
if settings.wikidata_enabled:
    
    pois_wd, wd = enrich_osm_with_wikidata(
        pois,
        settings.wikidata_user_agent,
        settings.wikidata_max_entities,
    )

else:

    pois_wd = pois.copy()
    wd = None

pois_wd.to_parquet(
    OUTPUT_PATH,
    index=False,
)

print("\n✓ Wikidata enrichment completed.")
print(f"Output rows: {len(pois_wd):,}")
print(
    f"Wikidata entities: "
    f"{0 if wd is None else len(wd):,}"
)
```

```python
print(
    "✓",
    OUTPUT_PATH,
    "exists:",
    OUTPUT_PATH.exists()
)
```
